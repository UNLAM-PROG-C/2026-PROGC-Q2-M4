```python
import time
import threading

def mostrar(accion, nombre, genero):
    transcurrido = time.perf_counter() - inicio

    with lock_impresion:
        print(
            f"[{transcurrido:5.1f}s] "
            f"{accion:<7} | "
            f"{nombre:<10} | "
            f"{genero}"
        )

lock_impresion = threading.Lock()
```

```python
def empleado(nombre, genero): 
    mostrar("ESPERA", nombre, genero)
    
    with lock_por_genero[genero]:
        cantidad_por_genero[genero] += 1
        if cantidad_por_genero[genero] == 1:
            baño_libre.acquire()
    
    capacidad_baño.acquire()
    mostrar("ENTRA", nombre, genero)
    
    time.sleep(5)            
    
    mostrar("SALE", nombre, genero)
    capacidad_baño.release()
    
    with lock_por_genero[genero]:
            cantidad_por_genero[genero] -= 1
            if cantidad_por_genero[genero] == 0:
                baño_libre.release()

empleados = [
    ("Ana", "mujer"),
    ("Juan", "hombre"),
    ("Lucía", "mujer"),
    ("Pedro", "hombre"),
    ("Sofía", "mujer"),
    ("Martín", "hombre"),
    ("Carla", "mujer"),
    ("Diego", "hombre"),
    ("Valentina", "mujer"),
    ("Nicolás", "hombre"),
    ("Camila", "mujer"),
    ("Federico", "hombre"),
    ("Julieta", "mujer"),
    ("Mateo", "hombre"),
    ("Agustina", "mujer"),
]

hilos = []

capacidad_baño = threading.BoundedSemaphore(3)
baño_libre = threading.Semaphore(1)
cantidad_por_genero = {
    "hombre": 0,
    "mujer": 0,
}
lock_por_genero = {
    "hombre": threading.Lock(),
    "mujer": threading.Lock(),
}

inicio = time.perf_counter()

for nombre, genero in empleados:
    hilo = threading.Thread(target=empleado, args=(nombre, genero))
    hilos.append(hilo)
    hilo.start()

for hilo in hilos:
    hilo.join()
```
